```python
# ============================================================
# ECON 425 Project:
# "Riding the AI Wave: Forecasting NVIDIA (NVDA) Returns"
# ============================================================
# Before running, install dependencies:
# pip install yfinance statsmodels arch matplotlib pandas numpy
# ============================================================

import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.regression.rolling import RollingOLS
import statsmodels.api as sm

from arch import arch_model

plt.rcParams["figure.figsize"] = (10, 6)

# ------------------------------------------------------------
# 1. Download data from Yahoo Finance (robust version)
# ------------------------------------------------------------

start_date = "2015-01-01"
end_date   = "2025-01-01"  # you can change this to today's date if you want

tickers = ["NVDA", "^GSPC", "^TNX"]
price_df_list = []

for t in tickers:
    tmp = yf.download(t, start=start_date, end=end_date)
    
    # Ensure tmp is a DataFrame
    if isinstance(tmp, pd.Series):
        tmp = tmp.to_frame()
    
    # Choose a usable price column
    if "Adj Close" in tmp.columns:
        s = tmp["Adj Close"]
    elif "Close" in tmp.columns:
        s = tmp["Close"]
    else:
        raise ValueError(
            f"No usable price column found for ticker {t}. "
            f"Columns available: {list(tmp.columns)}"
        )
    
    # Convert to DataFrame and name column as ticker
    if isinstance(s, pd.Series):
        s = s.to_frame(name=t)
    else:
        s = s.rename(columns={s.columns[0]: t})
    
    price_df_list.append(s)

# Combine all tickers into one DataFrame
data = pd.concat(price_df_list, axis=1)

# Clean column names (^GSPC, ^TNX → GSPC, TNX) and drop missing rows
data = data.rename(columns={"^GSPC": "GSPC", "^TNX": "TNX"})
data = data.dropna()

print("First few rows of price data:")
print(data.head())

# ------------------------------------------------------------
# 2. Construct returns
# ------------------------------------------------------------

# Log returns (×100 so roughly in percent)
returns = 100 * np.log(data / data.shift(1))
returns = returns.dropna()

# For TNX, also compute a simple change (level difference)
returns["TNX_change"] = data["TNX"].diff()
returns = returns.dropna()

print("\nFirst few rows of return data:")
print(returns.head())

# ------------------------------------------------------------
# 3. Descriptive statistics & basic plots
# ------------------------------------------------------------

print("\n=== Descriptive statistics for prices ===")
print(data.describe().T)

print("\n=== Descriptive statistics for returns ===")
print(returns[["NVDA", "GSPC", "TNX", "TNX_change"]].describe().T)

# NVDA price
plt.figure()
plt.plot(data.index, data["NVDA"], label="NVDA Price")
plt.title("NVDA Adjusted Closing Price")
plt.xlabel("Date")
plt.ylabel("Price ($)")
plt.legend()
plt.tight_layout()
plt.show()

# S&P 500 price
plt.figure()
plt.plot(data.index, data["GSPC"], label="S&P 500 Index")
plt.title("S&P 500 Index Level")
plt.xlabel("Date")
plt.ylabel("Index Level")
plt.legend()
plt.tight_layout()
plt.show()

# 10-year yield
plt.figure()
plt.plot(data.index, data["TNX"], label="10-Year Treasury Yield (^TNX)")
plt.title("10-Year Treasury Yield")
plt.xlabel("Date")
plt.ylabel("Yield (%)")
plt.legend()
plt.tight_layout()
plt.show()

# NVDA vs S&P returns
plt.figure()
plt.plot(returns.index, returns["NVDA"], label="NVDA Return")
plt.plot(returns.index, returns["GSPC"], label="S&P 500 Return", alpha=0.7)
plt.title("NVDA vs S&P 500 Daily Log Returns")
plt.xlabel("Date")
plt.ylabel("Return (%)")
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. Unit Root Tests (ADF)
# ------------------------------------------------------------

def adf_test(series, series_name="series"):
    print(f"\n=== ADF test for {series_name} ===")
    result = adfuller(series.dropna(), autolag="AIC")
    labels = ["ADF Statistic", "p-value", "# Lags Used", "# Observations Used"]
    out = pd.Series(result[0:4], index=labels)
    for key, val in result[4].items():
        out[f"Critical Value ({key})"] = val
    print(out)

# Prices
adf_test(data["NVDA"], "NVDA price")
adf_test(data["GSPC"], "GSPC price")
adf_test(data["TNX"],  "TNX level")

# Returns
adf_test(returns["NVDA"],       "NVDA return")
adf_test(returns["GSPC"],       "GSPC return")
adf_test(returns["TNX_change"], "TNX change")

# ------------------------------------------------------------
# 5. ARMA model selection for NVDA Returns
# ------------------------------------------------------------

y = returns["NVDA"].copy()

best_aic = np.inf
best_order = None
best_model = None

print("\n=== ARMA(p,q) model selection by AIC ===")
for p in range(0, 4):
    for q in range(0, 4):
        if p == 0 and q == 0:
            continue  # skip mean-only
        try:
            model = ARIMA(y, order=(p, 0, q))
            res = model.fit()
            print(f"ARMA({p},{q}) AIC: {res.aic:.2f}")
            if res.aic < best_aic:
                best_aic = res.aic
                best_order = (p, q)
                best_model = res
        except Exception as e:
            print(f"ARMA({p},{q}) failed: {e}")

print(f"\nBest ARMA order by AIC: {best_order}, AIC = {best_aic:.2f}")

print("\n=== Best ARMA model summary ===")
print(best_model.summary())

# ------------------------------------------------------------
# 6. Ex-post out-of-sample forecast (last 30 days)
# ------------------------------------------------------------

test_size = 30
y_train = y.iloc[:-test_size]
y_test  = y.iloc[-test_size:]

model_oos = ARIMA(y_train, order=(best_order[0], 0, best_order[1])).fit()
forecast_oos = model_oos.get_forecast(steps=test_size)
mean_pred = forecast_oos.predicted_mean
conf_int  = forecast_oos.conf_int(alpha=0.05)

# Align indices
mean_pred.index = y_test.index
conf_int.index  = y_test.index

# RMSFE
rmsfe = np.sqrt(np.mean((y_test - mean_pred) ** 2))
print(f"\nEx-post RMSFE over last {test_size} days: {rmsfe:.4f}")

# Plot actual vs forecast
plt.figure()
plt.plot(y_test.index, y_test, label="Actual NVDA Return")
plt.plot(mean_pred.index, mean_pred, label="Forecasted NVDA Return")
plt.fill_between(
    conf_int.index,
    conf_int.iloc[:, 0],
    conf_int.iloc[:, 1],
    alpha=0.2,
    label="95% Conf. Interval"
)
plt.title("Ex-Post Out-of-Sample Forecast: NVDA Returns")
plt.xlabel("Date")
plt.ylabel("Return (%)")
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. Ex-ante 5-day ahead forecast (using full sample)
# ------------------------------------------------------------

model_full = ARIMA(y, order=(best_order[0], 0, best_order[1])).fit()
forecast_5 = model_full.get_forecast(steps=5)
f_mean = forecast_5.predicted_mean
f_ci   = forecast_5.conf_int(alpha=0.05)

print("\n=== 5-day ahead ex-ante forecast for NVDA returns ===")
forecast_table = pd.DataFrame({
    "Forecast_Return": f_mean,
    "Lower_95": f_ci.iloc[:, 0],
    "Upper_95": f_ci.iloc[:, 1]
})
print(forecast_table)

# ------------------------------------------------------------
# 8. Advanced Model: GARCH(1,1) on NVDA Returns
# ------------------------------------------------------------

# Center y for GARCH
y_centered = y - y.mean()

garch_model = arch_model(y_centered, vol="Garch", p=1, q=1, dist="normal")
garch_res = garch_model.fit(update_freq=20, disp="off")

print("\n=== GARCH(1,1) results ===")
print(garch_res.summary())

# Conditional volatility
cond_vol = garch_res.conditional_volatility

plt.figure()
plt.plot(cond_vol.index, cond_vol, label="Conditional Volatility (sigma_t)")
plt.title("NVDA Conditional Volatility from GARCH(1,1)")
plt.xlabel("Date")
plt.ylabel("Volatility (Std Dev of Return)")
plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. Additional Advanced Model: Rolling AR(1) Regression
# ------------------------------------------------------------

window = 60  # rolling window length (days)

# Lagged NVDA returns
y_lag = y.shift(1).dropna()
y_rr  = y.loc[y_lag.index]

X_rr = sm.add_constant(y_lag)

rolling_model = RollingOLS(y_rr, X_rr, window=window)
rolling_res = rolling_model.fit()

beta_series = rolling_res.params["NVDA"]
const_series = rolling_res.params["const"]

# Plot rolling AR(1) coefficient
plt.figure()
plt.plot(beta_series.index, beta_series, label="Rolling AR(1) Coefficient")
plt.axhline(0, linestyle="--", linewidth=1)
plt.title(f"Rolling AR(1) Coefficient for NVDA Returns (Window = {window})")
plt.xlabel("Date")
plt.ylabel("Coefficient on lagged return")
plt.legend()
plt.tight_layout()
plt.show()

# Plot rolling intercept
plt.figure()
plt.plot(const_series.index, const_series, label="Rolling Intercept")
plt.axhline(0, linestyle="--", linewidth=1)
plt.title(f"Rolling Intercept for NVDA Returns (Window = {window})")
plt.xlabel("Date")
plt.ylabel("Intercept")
plt.legend()
plt.tight_layout()
plt.show()

print("\nScript complete. Use printed tables and plots for your slides.")
```
