# Question 2

Write a function called `row_means(M)` that takes a NumPy matrix and returns a NumPy array containing the mean of each row.

```python
def row_means(M):
    return M.mean(axis=1)
```

Test it on `M = np.array([[2, 4], [6, 8], [10, 12]])`

```python
M = np.array([[2, 4],
              [6, 8],
              [10, 12]])

row_means(M)
```

# Qurestion 5

Create a Pandas DataFrame called df_sales:

| day | sales |
| --- | ----- |
| Mon | 10    |
| Tue | 12    |
| Wed | 5     |
| Thu | 8     |
| Fri | 20    |


```python
sales = pd.DataFrame({
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "sales": [200, 220, 250, 210, 300]
})

sales
```

Compute the mean of the sales column:

```python
mean_sales = sales["sales"].mean()
mean_sales
```

Compute the maximum sale and what day it occurred on:

```python
max_sale = sales["sales"].max()
max_day = sales.loc[sales["sales"].idxmax(), "day"]

max_sale, max_day
```

# Question 6

Write a function called `center_column(series)` that takes a Pandas Series and returns a centered version:

$$x_c = x - mean(x)$$

```python
def center_column(series):
    return series - series.mean()
```

Test it on `s = pd.Series([10, 20, 30, 40])`

```python
s = pd.Series([10, 20, 30, 40])
center_column(s)
```

# Question 7

Create the following DataFrame:

| team | points |
| ---- | ------ |
| A    | 10     |
| B    | 20     |
| A    | 15     |
| B    | 18     |
| A    | 30     |


```python
points = pd.DataFrame({
    "team": ["A", "B", "A", "B", "A"],
    "points": [10, 20, 15, 18, 30]
})

points
```

Using `.groupby()`, compute total points per team

```python
points.groupby("team")["points"].sum()
```

Using `.groupby()` compute average points per team

```python
points.groupby("team")["points"].mean()
```

Using `.groupby()`, produce A DataFrame sorted by total points (descending)

```python
total_points = points.groupby("team")["points"].sum().reset_index()
total_points.sort_values("points", ascending=False)
```
