```python
import numpy as np
np.array([1,2,3])                  # from list
np.zeros(5)                        # 1D zeros
np.ones((2,3))                     # 2x3 ones
np.full((3,3), 7)                  # filled with 7
np.empty((2,2))                    # uninitialized (don’t rely on values)
np.arange(0, 10, 2)                # 0..8 step 2
np.linspace(0, 1, 5)               # 5 evenly spaced in [0,1]
```
