```python
import numpy as np
import matplotlib.pyplot as plt
from random import choice
```

```python
step_fun = lambda x: x > 0
x = np.linspace(-10,10,100)
z = step_fun
#plt.plot(x,z)
#plt.ylim()
```

```python
training_data = [
    (np.array([0,0,1]), 1),
    (np.array([0,1,1]), 0),
    (np.array([1,0,1]), 0),
    (np.array([1,1,1]), 0),
]

# usu. random numbers for weights is not a bad starting point

w = np.random.rand(3)*10
errors = []
ws0 = []
ws1 = []
ws2 = []

# Learning rate
alpha = 0.3 # if w initial values are given a range 10x larger, use large alpha and 200 steps

# use 100 steps
n = 100

#w. history = []
for i in range(n):
    #stochaistic
    x, target = choice(training_data)

    y= np.dot(w,x)
    z = step_fun(y)

    #backward propogation
    error = target - z
    errors.append(error)
    w += alpha * error * x
    #error * x is the gradient/derivative

    ws0.append(w[0])
    ws1.append(w[1])
    ws2.append(w[2])

print('Trained Weights: {}\n'.format(w))

for x, _ in training_data:
    y = np.dot(x,w)
    print("{}: {:.4f} -> {:d}".format(x[:2], y, step_fun(y)))
```

```python
# Breakout: 20 random pairs in [0,1], classify by z, and plot
rng = np.random.default_rng(0)
pairs = rng.random((20, 2))
X = np.hstack([pairs, np.ones((20, 1))])
y = X @ w
z = step_fun(y).astype(int)

for p, yi, zi in zip(pairs, y, z):
    print(f"{p}: {yi:.4f} -> {zi}")

colors = np.where(z == 1, 'tab:blue', 'tab:red')
plt.scatter(pairs[:, 0], pairs[:, 1], c=colors, s=60, edgecolors='k')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Random Pairs Classified by z')
plt.grid(True, alpha=0.3)
from matplotlib.lines import Line2D
legend_elems = [
    Line2D([0], [0], marker='o', color='w', label='z=1', markerfacecolor='tab:blue', markersize=8, markeredgecolor='k'),
    Line2D([0], [0], marker='o', color='w', label='z=0', markerfacecolor='tab:red', markersize=8, markeredgecolor='k'),
]
plt.legend(handles=legend_elems, loc='best')
plt.show()
```
