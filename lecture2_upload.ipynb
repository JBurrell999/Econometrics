# Vectorization

In general operations like `+`, `-`, `*`, `/` and `**` act elementwise on arrays.

So if `X` is a NumPy array, `X**2` squares each element. 

This is **vectorization**. Basically it means you can do stuff on an array ("vector") all at once, instead of looping through the array and doing the stuff on one element at a time.

How? 

NumPy still uses loops -- it's just all under the hood. An operation like `**` gets sent to a "subroutine" written in C -- a much faster language than Python -- that executes the loop. 

(For those of who come from R, this seems obvious, because R is vectorized by default, unlike Python.)

# Random numbers

**Econometrics and data science are about learning and making predictions in the face of this uncertainty.** A random variable is just a numerical way to describe uncertainty, and probability distributions tell us how likely each outcome is. 

Probability distributions are fundamendamental in econometrics and data science. Probably the most common use of probbility distribution is statistical inference. We often need to make inference about population based on sample data. By making distributional assumptions on data and modeling relationships between variables as random variables with specific distributions, we can make probabilistic statements about the parameters. 

A probability distribution is a mathematical function that describes the likelihood of various outcomes or events in a random experiment or process. It assigns probabilities to each possible outcome, indicating how likely each outcome is to occur. 

When we analyze data, we often face uncertainty. Prices change, survey answers differ, stock returns fluctuate, and test scores vary across people. A random variable is simply a way to describe this uncertainty using numbers.

Think of a random variable as a rule that assigns a number to the outcome of some random process:

- The roll of a die → numbers 1 through 6

- Daily return of a stock → any real number (positive or negative)

- Number of job offers someone gets → 0, 1, 2, …

A probability distribution tells us how likely different values of that random variable are. It’s like a map of uncertainty: it assigns probabilities to each possible outcome.

- For a die roll: each of 1–6 has probability 1/6 (uniform distribution).

- For stock returns: most days are close to zero, but large ups or downs are possible (often modeled with a bell-shaped distribution).

- For the number of job offers: probabilities are higher for 0 or 1, lower for very large numbers.

Probability distribution can be either discrete or continuous.

- Discrete random variable: can only take countable number of distinct outcomes.
- Continuous random variable: outcome takes the form of continuous range.

As you will see, as we get to estimations, we will assume distributional assumptions to model enviroments and to estimate parameters of interest. But sometimes we want to simulate data itself to validate methodological choices, estimate complex models, improve experiment design etc. 


Numpy provides some functions to simulate random variables from different distributions.

## Normal Distribution

$X \sim N(\mu, \sigma^2)$ means "X is normally distribution with a mean of $\mu$ and variance of $\sigma^2$."

The function in Numpy is `np.random.normal(loc, scale, size)` where:

* `loc` is the mean`
* `scale` is the **standard deviation** $\sigma$ (not the variance -- standard deviation is just the square root of the variance)
* `size` is the number of observations to generate
