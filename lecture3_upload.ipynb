Series are the equivalent of 1-dimensional arrays. DataFrames are collections of Series and so are 2-dimensional. Panels are collections of DataFrames, and have 3 dimensions.

### Pandas DataFrame

DataFrames collect multiple series in the same way that a spreadsheet collects multiple columns of data. In
a simple sense, a DataFrame is like a 2-dimensional NumPy array – and when all data is numeric and of the
same type (e.g. float64), it is virtually indistinguishable. However, a DataFrame is composed of Series and
each Series has its own data type, and so not all DataFrames are representable as homogeneous NumPy arrays.

### Summary (Panda)

Other important pandas methods that you will use a lot and learn as you use python more and more, so you don't need to remmember them, just google when you need to do something with the dataframe and you will find them

- unique() returns the unique elements of a series
- nunique() returns the number of unique values in a Series
- drop(labels) drop elements with the selected labels from a Series.
- dropna() is similar to drop() except that it only drops null values – NaN or similar
- fillna(value) fills all null values in a series with a specific value.
- merge() and join() provide SQL-like operations for merging the DataFrames using row labels or the contents of columns. The primary difference between the two is that merge defaults to using column contents while join defaults to using index labels.
- pd.concat() is a core function which concatenates two or more DataFrames using an outer join by default.
- sort_values() and sort_index() provide methods to sort a DataFrame. sort_values sorts the contents of the DataFrame along either axis using the contents of a single column or row. Passing a list of coumns names or index values implements lexicographic search. sort_index will sort a DataFrame by the values in the index.
- append() appends rows of another DataFrame to the end of an existing DataFrame
- apply() executes a function along the columns or rows of a DataFrame.
- pivot_table() provides a method to summarize data by groups.
