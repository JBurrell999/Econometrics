## Machine Learning Metrics and Bayes’ Theorem

This notebook answers **Problem A (parts a & b)** from Homework 1.

**Given (from the PDF):**
- Validation set negatives: **N = 6300**
- Validation set positives: **P = 190**
- Precision: **p = 0.83**
- Recall: **r = 0.87**
- Real-world prevalence (base rate): **1 per 10,000 images**

**Definitions (from the PDF):**
\[
\text{Precision } p = \frac{TP}{TP+FP}\qquad \text{Recall } r = \frac{TP}{TP+FN} = \frac{TP}{P}
\]

We will compute:
- **TPR** and **FPR** for part (a)
- Expected number of images to inspect per one true microorganism hit for part (b)


## Part (a): Compute TPR and FPR

### Step 1 — From recall, compute TP and FN
Recall is
\[
 r = \frac{TP}{P}
\]
so
\[
 TP = r\,P,\qquad FN = P-TP.
\]

### Step 2 — From precision, compute FP
Precision is
\[
 p = \frac{TP}{TP+FP}
\]
so
\[
 FP = TP\left(\frac{1}{p}-1\right).
\]

### Step 3 — Compute TPR and FPR
\[
TPR = \frac{TP}{P} = r\qquad\text{and}\qquad FPR = \frac{FP}{N}.
\]


```
# Inputs from the problem
N = 6300   # number of negatives in validation
P = 190    # number of positives in validation
p = 0.83   # precision
r = 0.87   # recall

# Compute counts
TP = r * P
FN = P - TP
FP = TP * (1/p - 1)

# Rates
TPR = TP / P
FPR = FP / N

TP, FN, FP, TPR, FPR
```

### Rounded confusion-matrix counts (optional)
In practice, counts should be integers. Because precision/recall were reported with two decimals, rounding is reasonable.


```
TP_i = int(round(TP))
FN_i = int(round(FN))
FP_i = int(round(FP))
TN_i = N - FP_i

TP_i, FN_i, FP_i, TN_i
```

## Part (b): Expected number of images to inspect per one true microorganism

You are told the microorganism occurs at **1 per 10,000** images.

Consider a large batch of **10,000** images:
- Expected **true positives** flagged by the model:
\[
\mathbb{E}[TP_{\text{flagged}}] = TPR \times 1
\]
- Expected **false positives** flagged by the model:
\[
\mathbb{E}[FP_{\text{flagged}}] = FPR \times 9999
\]

Total images you would inspect (those flagged positive) per 10,000 images:
\[
\mathbb{E}[\text{flagged}] = \mathbb{E}[TP_{\text{flagged}}] + \mathbb{E}[FP_{\text{flagged}}]
\]

Then the expected number of inspected images **per one true microorganism** is approximately:
\[
\frac{\mathbb{E}[\text{flagged}]}{\mathbb{E}[TP_{\text{flagged}}]}.
\]


```
# Base rate information
total = 10_000
true_pos_rate = 1/10_000

expected_actual_pos = total * true_pos_rate          # = 1
expected_actual_neg = total - expected_actual_pos    # = 9999

expected_true_flagged = TPR * expected_actual_pos
expected_false_flagged = FPR * expected_actual_neg
expected_flagged_total = expected_true_flagged + expected_false_flagged

# Images inspected per one true microorganism (approx)
inspected_per_true = expected_flagged_total / expected_true_flagged

expected_actual_pos, expected_true_flagged, expected_false_flagged, expected_flagged_total, inspected_per_true
```

### Final numeric summary


```
import pandas as pd

summary = pd.DataFrame([
    {
        'TPR (a.k.a. Recall)': TPR,
        'FPR': FPR,
        'Expected true flagged per 10,000': expected_true_flagged,
        'Expected false flagged per 10,000': expected_false_flagged,
        'Expected flagged total per 10,000': expected_flagged_total,
        'Expected inspected images per 1 true microorganism': inspected_per_true,
    }
])
summary
```
