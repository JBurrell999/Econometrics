## $$ z = tanh(y)$$

## Intro:

## $$sinh(y) = \frac {e^y - e^{-y}} {2}$$
## $$cosh(y) = \frac {e^y + e^{-y}} {2}$$
## $$tanh(y) = sinh(y)/cosh(y)$$

- ## A) Choose a reasonable range of $y$ and plot sinh, cosh in one figure, and tanh in another.  Do they look like what you would expect?
- ## B) Find the derivative of $tanh(y)$.  First express it as a function of $y$. 
- ## C) Then express it as a function of $z$.
- ## D) Compute the numerical derivative of $tanh(y)$ between y = -5 and +5.
- ## E) Plot the numerical derivative you got in Part D), and on the same figure, plot the analytical derivative you found in Part C) -- do they agree?

```python
# Part A: Plot sinh, cosh in one figure, tanh in another

y = np.linspace(-5, 5, 200)

#sinh and cosh
plt.plot(y, np.sinh(y), label='sinh(y)')
plt.plot(y, np.cosh(y), label='cosh(y)')
plt.title('sinh and cosh')
plt.legend()
plt.show()

#tanh
plt.plot(y, np.tanh(y), label='tanh(y)')
plt.title('tanh')
plt.legend()
plt.show()
```

## Part B: Derivative of tanh(y) as a function of y
## Using the quotient rule on sinh/cosh:
## $$\frac{d}{dy}\tanh(y) = \frac{\cosh^2(y) - \sinh^2(y)}{\cosh^2(y)} = \frac{1}{\cosh^2(y)}$$

## Part C: As a function of z
## Since $z = \tanh(y)$ and $\tanh^2(y) = 1 - \frac{1}{\cosh^2(y)}$:
## $$\frac{dz}{dy} = 1 - \tanh^2(y) = 1 - z^2$$

```python
# Part D: derivative of tanh(y) [ -5 and +5 ]
y = np.linspace(-5, 5, 200)
tanh = np.tanh(y)

# derivative he wrote out for us dy/dx ≈ (f(x+h) - f(x)) / h
dy = np.diff(tanh)
dx = np.diff(y)
numerical_deriv = dy / dx
y_mid = (y[:-1] + y[1:]) / 2  # midpoints

# Part E: plotted
# recall 1 - z^2 in c
analytical_deriv = 1 - np.tanh(y)**2

plt.figure(figsize=(6, 4))
plt.plot(y_mid, numerical_deriv, 'b.', label='num derivative', markersize=3)
plt.plot(y, analytical_deriv, 'r-', label='Analytical: $1 - z^2$')
plt.title('Numerical vs Analytical Derivative of tanh(y)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
