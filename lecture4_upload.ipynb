## Vector Operations


Vector in general is defined as a quantity having direction as well as magnitude. But for the purpose of this course we refer a vector to be sequence of numbers. For example [1, 2, 40, 3] is a vector with length 4. When programming, we can use list, numpy array, or pandas series to represent vector. There are few operations that we need to know. 

- **Vector addition**: if we have two vectors $a$ and $b$ of length $n$ (we will write this as $a, b \in \mathbb{R}^n$), we can add them element by element:

$$ a + b = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix} + \begin{bmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{bmatrix} = \begin{bmatrix} a_1+b_1 \\ a_2+b_2 \\ \vdots \\ a_n+b_n\end{bmatrix}$$

- **Scalar multiplication**: if we have a vector $a\in \mathbb{R}^n$ and scalar $\theta\in \mathbb{R}$: 

$$ \theta \cdot a = \theta \cdot \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_n \end{bmatrix} = \begin{bmatrix} \theta \cdot a_1 \\ \theta \cdot a_2 \\ \vdots \\ \theta \cdot a_n\end{bmatrix}$$

- **Inner (dot) product** : the inner product of two vectors is a measure of how aligned they are. Mathematically, if $a, b \in \mathbb{R}^n$, inner product is calculated as:

$$ a\cdot b=a^T b = [a_1 a_2 \cdots a_n]\begin{bmatrix}b_1 \\ b_2 \\ \vdots \\ b_n\end{bmatrix} = a_1\cdot b_1+a_2\cdot b_2+\cdots+a_n\cdot b_n = \sum_{i=1}^n a_i\cdot b_i $$

We denote **transpose** by $T$ or by $'$. Traspose is operation that flips given matrix (or vector in this case) by making rows as columns and columns as rows. 



- **Norm**: is a length of a vector. There are few ways people calculate it but the most popular (and intuative) is euclidean norm (also called L2 norm). If we have a vector $a\in \mathbb{R}^n$:

$$||a||=\sqrt{a^Ta}=\sqrt{\sum_{i=1}^n a_i^2}$$

## Matrix Operations


In mathematics, a matrix is a rectangular array or table of **numbers**, symbols, or expressions, arranged in rows and columns. This means that if in case of vector we have only one subscript. Now, we have two subscripts, one inicating the row and another indicating the column. For example matrix of size $n \times m$ will look something like this: 

$$ X = \begin{bmatrix} x_{11} & x_{12} & \cdots & x_{1m} \\ x_{21} & x_{22} & \cdots & x_{2m}  \\ \vdots & \vdots & \ddots & \vdots \\ x_{n1} & x_{n2} & \cdots & x_{nm}  \end{bmatrix} $$

We say element $x_{ij}$ is an element of $X$ which is located at the interection of $i^{th}$ row and $j^{th}$ column. 

Scalar maltiplication and addition (matrix to matrix addition) works the same way as in case of vectors so we don't need to spend any time on that. Let's discuss matrix multiplication. It is kind of extension of inner product discussed above. 

Let's say we have two matrices $X$ and $Y$. In order X and Y to be "multiplyable" to each other, the number of columns in matrix X should be the same as the number of rows in matrix Y. That is, if the size of the matrix $X$ is $n \times m$, the size of matix $Y$ must be $m \times k$, where $k$ could be anything. To make notation clear, lets assume $X$ is $2\times3$ matrix and $Y$ is $3\times4$ matrix. Since the number of columns of $X$ equals to the number of rows of $Y$, we can calculate $XY$:

$$ X_{2\times3} \cdot Y_{3\times4} = \begin{bmatrix} x_{11} & x_{12} & x_{13} \\ x_{21} & x_{22} & x_{23} \end{bmatrix}_{2\times3} \begin{bmatrix} y_{11} & y_{12} & y_{13} & y_{14} \\ y_{21} & y_{22} & y_{23} & y_{24} \\ y_{31} & y_{32} & y_{33} & y_{34} \end{bmatrix}_{3\times4} = \begin{bmatrix} z_{11} & z_{12} & z_{13} & z_{14} \\ z_{21} & z_{22} & z_{23} & z_{24} \end{bmatrix}_{2\times4} $$

where:

$$z_{ij}= x_{i1} \cdot y_{1j} + x_{i2}\cdot y_{2j} + x_{i3} \cdot y_{3j}$$


That is to get an element $z_{ij}$, we should take $i^{th}$ row from $X$ and $j^{th}$ column from $Y$ and perform dot product opperation. Notice, the resulting matrix will have the the same number of rows as $X$ and the same number of columuns as $Y$.

## Solving linear equations

Let's assume we have the following demand equations for two related goods, say cereal (denote by good 0) and granola (good 1). The demand for both goods are represented as:

$$ q_0^d = 100 - 10p_0 - 5p_1 $$
$$ q_1^d = 50 - p_0 - 10p_1 $$

Also, supply on both markets are given as:

$$ q_0^s = 10p_0 + 5p_1 $$
$$ q_1^s = 5p_0 + 10p_1 $$

The equilibrium condition would be when the deamand on both markets equals to respective supplies. Therfore we have the folowing system of linear equations to solve:

$$ 100 - 10p_0 - 5p_1 = 10p_0 + 5p_1 $$
$$ 50 - p_0 - 10p_1 = 5p_0 + 10p_1  $$

How do we solve this? When it is such an easy example, we can always just solve it on a paper using "substitution" method. But imagine having many goods. Let's show how we can solve it using matrix algebra. For that we need to rewrite demand and supply equations into matrix form:

$$ q^d = Dp + h  \quad \text{where} \quad  q^d = \begin{bmatrix} q_0^d \\ q_1^d \end{bmatrix} \quad D = \begin{bmatrix} -10 & -5 \\ -1 & -10 \end{bmatrix} \quad \text{and} \quad \begin{bmatrix} 100 \\ 50 \end{bmatrix} $$ 


Similarly, we can rewrite supply:

$$ q^s = Cp  \quad \text{where} \quad  q^s = \begin{bmatrix} q_0^s \\ q_1^s \end{bmatrix} \quad C = \begin{bmatrix} 10 & 5 \\ 1 & 10 \end{bmatrix}$$ 

Now, we are looking at the solution for $q^s=q^d$:

$$ Cp = Dp + h $$

After rearanging terms we get:

$$ (C - D)p = h $$

Here $C$ and $D$ are matrices while $p$ and $h$ are vectors. So we can not simply do $p=h/(C-D)$. Hovewer, matrices allow something similar by taking inverse of a matrix. In this case we can rewrite as:

$$ p = (C - D)^{-1}h$$

**Hovewer**, this might not always be possible to do. The matrix is invertable if and only if the determinant is nonzero. In this case we say the matrix is nonsingular or rows and columns are linearly independant. Here is an example singular matrix:

$$\begin{bmatrix} 1 & 2 \\ 2 & 4 \end{bmatrix}$$

As you see the second row is just two multiplied the first row. Generally, if you can find any row or column that is a linear combination of any other rows or columns we say that A is linearly dependant and is not invertable. In other words, a square matrix $X$ is nonsingular if and only if the rows and columns of $X$ are linearly independent.

# Linear optimization problems

Linear programming problems optimize linear objective functions subject to a set of some linear constaints. Let's say we are economist in some company and our goal is to maximize revenue. Our company can produce two goods, good 1 and good 2:

- To produce one unit of good 1 we need 2 units of material and 4 units of labor and th price for good 1 is 3
- To produce one unit of good 2 we need 5 units of material and 2 units of labor and th price for good 2 is 4

30 units of material and 20 units of labor are available to us. Let's write this as a constrained optimization problem. Denote decision variable (how many units of good 1 and good 2 to produce respectively) by $x_1$ and $x_2$. We can write our optimization problem as:

\begin{aligned}
\max_{x_1,x_2} \ & 3 x_1 + 4 x_2 \ & \text{Total revenue} \\
\mbox{subject to } \ & 2 x_1 + 5 x_2 \le 30 \ & \text{Material constraint} \\
& 4 x_1 + 2 x_2 \le 20 \ & \text{labor constraint}\\
& x_1, x_2 \ge 0 \ & \text{non-negativity constraint} \\
\end{aligned}


Let's use scipy module to solve such constrained optimization problem.

```python
#import linprog module from scipy

from scipy.optimize import linprog
```

The linprog package solves problems of the following form (see the [full documantation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.linprog.html#scipy.optimize.linprog)):

\begin{aligned}
\max_{x} \ & c^Tx \\
\mbox{subject to } \ & A_{ub}x \le b_{ub}  \\
& A_{eq}x = b_{eq} \\
& l \le x \le u  \\
\end{aligned}

Where:
- $x$ is a vector of decision variables
- $c$ is the coefficients of the linear objective function to be minimized
- $A_{ub}$ The inequality constraint matrix. Each row of $A_{ub}$ specifies the coefficients of a linear inequality constraint on $x$.
- $b_{ub}$ The inequality constraint vector. Each element represents an upper bound on the corresponding value of inner product of $A_{ub}$ and $x$
- $A_{eq}$ The equality constraint matrix. Each row of $A_{eq}$ specifies the coefficients of a linear equality constraint on $x$.
- $b_{eq}$ The equality constraint vector. Each element of inner product of $A_{eq}$ and $x$ must equal the corresponding element of $b_{eq}$.
- ($l$, $u$) the lower and upper bounds of each element of decision variable $x$. By default, $l=0$ and $u=None$
