```python
# %%
# ==========================================================
# NeurIPS-style RF & XGBoost on LSMS + EE_harvest
# ==========================================================

import os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from tqdm import tqdm

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GroupKFold, ParameterGrid
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

from xgboost import XGBRegressor
import joblib

RANDOM_STATE = 42

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

print("Imports complete.")

# %%
# ==========================================================
# 1. Paths & Data Loading
# ==========================================================

LSMS_PATH = "/Users/jjburrell/Downloads/send/lsms_panel_allvars_2010_2024_MASTER.csv"
EE_PATH   = "/Users/jjburrell/Downloads/send/EE_harvest_ml.csv"

print("🔹 Loading LSMS from:", LSMS_PATH)
lsms_df = pd.read_csv(LSMS_PATH)
print("LSMS shape:", lsms_df.shape)
print("LSMS columns:", lsms_df.columns.tolist())
display(lsms_df.head())

print("\n🔹 Loading EE (harvest features) from:", EE_PATH)
ee_df = pd.read_csv(EE_PATH)
print("EE shape:", ee_df.shape)
print("EE columns:", ee_df.columns.tolist())
display(ee_df.head())

# %%
# ==========================================================
# 2. Clean LSMS and Build Target (NDVI per location)
# ==========================================================

required_cols = ["ea_id_obs", "ea_id_merge", "lat_modified", "lon_modified", "NDVI"]
missing = [c for c in required_cols if c not in lsms_df.columns]
if missing:
    raise ValueError(f"Missing required LSMS columns: {missing}")

# Drop rows with missing IDs/NDVI
lsms_df = lsms_df.dropna(subset=required_cols)
print("LSMS after dropping NA in ID/NDVI:", lsms_df.shape)

# Aggregate NDVI per location (ea_id + lat/lon)
group_keys = ["ea_id_obs", "ea_id_merge", "lat_modified", "lon_modified"]
agg_lsms = (
    lsms_df
    .groupby(group_keys, dropna=False)
    .agg(target_ndvi=("NDVI", "mean"))
    .reset_index()
)

print("Aggregated LSMS NDVI shape:", agg_lsms.shape)
display(agg_lsms.head())

# %%
# ==========================================================
# 3. Merge LSMS target with EE features
# ==========================================================

merge_keys = ["ea_id_obs", "ea_id_merge", "lat_modified", "lon_modified"]
for k in merge_keys:
    if k not in ee_df.columns:
        raise ValueError(f"EE is missing merge key column: {k}")

print("Merging LSMS NDVI with EE features on:", merge_keys)

df = pd.merge(
    agg_lsms,
    ee_df,
    on=merge_keys,
    how="inner"
)

df = df.dropna()
print("✅ Final merged dataset shape:", df.shape)
display(df.head())

merged_path = "merged_lsms_ee_ndvi.csv"
df.to_csv(merged_path, index=False)
print("Saved merged dataset to:", merged_path)

# %%
# ==========================================================
# 4. Features, Target, Groups, Train/Test Split (by wave)
# ==========================================================

TARGET_COL = "target_ndvi"

id_cols = [
    "country", "wave", "season",
    "ea_id_obs", "ea_id_merge",
    "lat_modified", "lon_modified"
]
id_cols = [c for c in id_cols if c in df.columns]

drop_cols = id_cols + [TARGET_COL]
feature_cols = [c for c in df.columns if c not in drop_cols]

print("Target:", TARGET_COL)
print("Num features:", len(feature_cols))
print("Example feature cols:", feature_cols[:15])

X_full = df[feature_cols]          # keep original index
y_full = df[TARGET_COL]
groups_full = df["ea_id_merge"]

if "wave" not in df.columns:
    raise ValueError("Column 'wave' not found in merged df; required for train/test split.")

max_wave = int(df["wave"].max())
train_mask = df["wave"] < max_wave
test_mask  = df["wave"] == max_wave

X_train = X_full.loc[train_mask].copy()
y_train = y_full.loc[train_mask].copy()
groups_train = groups_full.loc[train_mask].copy()

X_test  = X_full.loc[test_mask].copy()
y_test  = y_full.loc[test_mask].copy()
meta_test = df.loc[test_mask].copy()

print(f"Train waves: {sorted(df.loc[train_mask, 'wave'].unique())}")
print(f"Test wave: {max_wave}")
print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
print("Train countries:", df.loc[train_mask, 'country'].unique())
print("Test countries:", df.loc[test_mask, 'country'].unique())

# %%
# ==========================================================
# 5. CV Setup + Hyperparameter Grids (RF & XGB)
# ==========================================================

cv = GroupKFold(n_splits=5)

rf_param_grid = {
    "n_estimators": [300, 600],
    "max_depth": [12, 24, None],
    "max_features": [0.5, 0.8, 1.0],
    "min_samples_split": [2, 8],
    "min_samples_leaf": [1, 4],
    "bootstrap": [True],
}

xgb_param_grid = {
    "n_estimators": [400, 800],
    "max_depth": [4, 6, 8],
    "learning_rate": [0.03, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0],
    "reg_lambda": [1.0, 5.0],
}

rf_grid_list  = list(ParameterGrid(rf_param_grid))
xgb_grid_list = list(ParameterGrid(xgb_param_grid))

print("RF combos:", len(rf_grid_list))
print("XGB combos:", len(xgb_grid_list))

# %%
# ==========================================================
# 6. Spatial CV Tuning Helper (no early stopping)
# ==========================================================

def spatial_cv_tuning(
    model_name,
    model_class,
    base_kwargs,
    grid_list,
    X,
    y,
    groups,
    cv,
):
    best_r2 = -np.inf
    best_params = None

    for params in tqdm(grid_list, desc=f"Tuning {model_name}", ncols=100):
        r2_folds = []

        for train_idx, val_idx in cv.split(X, y, groups):
            X_tr, X_va = X.iloc[train_idx], X.iloc[val_idx]
            y_tr, y_va = y.iloc[train_idx], y.iloc[val_idx]

            model = model_class(**base_kwargs, **params)
            model.fit(X_tr, y_tr)

            y_pred = model.predict(X_va)
            r2 = r2_score(y_va, y_pred)
            r2_folds.append(r2)

        r2_mean = float(np.mean(r2_folds))

        if r2_mean > best_r2:
            best_r2 = r2_mean
            best_params = params

    print(f"✅ {model_name} best CV R²={best_r2:.3f}")
    print("Best params:", best_params)
    return best_params, best_r2

# %%
# ==========================================================
# 7. Tune + Train GLOBAL RF & XGB on TRAIN
# ==========================================================

rf_base_kwargs = {
    "n_jobs": -1,
    "random_state": RANDOM_STATE,
}

xgb_base_kwargs = {
    "n_jobs": -1,
    "tree_method": "hist",
    "random_state": RANDOM_STATE,
}

print("\n🌍 Tuning GLOBAL Random Forest on TRAIN...")
best_rf_params, rf_cv_r2 = spatial_cv_tuning(
    model_name="RF",
    model_class=RandomForestRegressor,
    base_kwargs=rf_base_kwargs,
    grid_list=rf_grid_list,
    X=X_train,
    y=y_train,
    groups=groups_train,
    cv=cv,
)

print("\n🌍 Tuning GLOBAL XGBoost on TRAIN...")
best_xgb_params, xgb_cv_r2 = spatial_cv_tuning(
    model_name="XGB",
    model_class=XGBRegressor,
    base_kwargs=xgb_base_kwargs,
    grid_list=xgb_grid_list,
    X=X_train,
    y=y_train,
    groups=groups_train,
    cv=cv,
)

global_rf = RandomForestRegressor(**rf_base_kwargs, **best_rf_params)
global_rf.fit(X_train, y_train)

global_xgb = XGBRegressor(**xgb_base_kwargs, **best_xgb_params)
global_xgb.fit(X_train, y_train)

print("\n✅ Fitted GLOBAL RF and XGBoost on TRAIN")

# %%
# ==========================================================
# 8. Evaluation Helper
# ==========================================================

def evaluate_model(model, X, y):
    y_pred = model.predict(X)
    r2 = r2_score(y, y_pred)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    mae = mean_absolute_error(y, y_pred)
    return r2, rmse, mae, y_pred

# %%
# ==========================================================
# 9. Country-Specific vs Global (RF & XGB) – Training & Results
# ==========================================================

countries = sorted(df["country"].unique())

rows_rf = []
rows_xgb = []

country_models_rf = {}
country_models_xgb = {}

for c in countries:
    train_mask_c = (df["country"] == c) & train_mask
    test_mask_c  = (df["country"] == c) & test_mask

    Xc_train = X_full.loc[train_mask_c]
    yc_train = y_full.loc[train_mask_c]
    groups_c_train = groups_full.loc[train_mask_c]

    Xc_test = X_full.loc[test_mask_c]
    yc_test = y_full.loc[test_mask_c]

    print(f"\n=== {c} ===")
    print(f"  Train n={len(Xc_train)}, Test n={len(Xc_test)}")

    if len(Xc_test) == 0:
        print("  ⚠️ No test data for this country in held-out wave; skipping.")
        continue

    # Global models performance
    r2_rf_glob, _, _, _ = evaluate_model(global_rf, Xc_test, yc_test)
    r2_xgb_glob, _, _, _ = evaluate_model(global_xgb, Xc_test, yc_test)

    rows_rf.append({"country": c, "model": "Global RF", "r2": r2_rf_glob})
    rows_xgb.append({"country": c, "model": "Global XGB", "r2": r2_xgb_glob})

    if len(Xc_train) < 100:
        print("  ⚠️ Not enough train data for country-specific models, using global only.")
        continue

    # Country-specific RF
    print(f"  🔧 Tuning country-specific RF for {c} ...")
    best_rf_c, cv_r2_rf_c = spatial_cv_tuning(
        model_name=f"RF_{c}",
        model_class=RandomForestRegressor,
        base_kwargs=rf_base_kwargs,
        grid_list=rf_grid_list,
        X=Xc_train.reset_index(drop=True),
        y=yc_train.reset_index(drop=True),
        groups=groups_c_train.reset_index(drop=True),
        cv=cv,
    )

    rf_c = RandomForestRegressor(**rf_base_kwargs, **best_rf_c)
    rf_c.fit(Xc_train, yc_train)
    country_models_rf[c] = rf_c

    r2_rf_c, _, _, _ = evaluate_model(rf_c, Xc_test, yc_test)
    rows_rf.append({"country": c, "model": "Country-Specific RF", "r2": r2_rf_c})

    # Country-specific XGB
    print(f"  🔧 Tuning country-specific XGB for {c} ...")
    best_xgb_c, cv_r2_xgb_c = spatial_cv_tuning(
        model_name=f"XGB_{c}",
        model_class=XGBRegressor,
        base_kwargs=xgb_base_kwargs,
        grid_list=xgb_grid_list,
        X=Xc_train.reset_index(drop=True),
        y=yc_train.reset_index(drop=True),
        groups=groups_c_train.reset_index(drop=True),
        cv=cv,
    )

    xgb_c = XGBRegressor(**xgb_base_kwargs, **best_xgb_c)
    xgb_c.fit(Xc_train, yc_train)
    country_models_xgb[c] = xgb_c

    r2_xgb_c, _, _, _ = evaluate_model(xgb_c, Xc_test, yc_test)
    rows_xgb.append({"country": c, "model": "Country-Specific XGB", "r2": r2_xgb_c})

rf_comp_df = pd.DataFrame(rows_rf)
xgb_comp_df = pd.DataFrame(rows_xgb)

print("\nRandom Forest comparison:")
display(rf_comp_df.sort_values(["country", "model"]))

print("\nXGBoost comparison:")
display(xgb_comp_df.sort_values(["country", "model"]))

# %%
# ==========================================================
# 10. Plots: Country-Specific vs Global R² and Save Artifacts
# ==========================================================

model_dir = "./models_neurips_ready"
os.makedirs(model_dir, exist_ok=True)

# Save global models
joblib.dump(global_rf, os.path.join(model_dir, "global_rf_lsms_ee_ndvi_waveholdout.joblib"))
joblib.dump(global_xgb, os.path.join(model_dir, "global_xgb_lsms_ee_ndvi_waveholdout.joblib"))

# Save comparison tables
rf_path = os.path.join(model_dir, "rf_country_vs_global.csv")
xgb_path = os.path.join(model_dir, "xgb_country_vs_global.csv")
rf_comp_df.to_csv(rf_path, index=False)
xgb_comp_df.to_csv(xgb_path, index=False)

print("Saved RF comparison to:", rf_path)
print("Saved XGB comparison to:", xgb_path)

# Random Forest comparison plot
plt.figure(figsize=(12, 4))
sns.barplot(
    data=rf_comp_df,
    x="country",
    y="r2",
    hue="model",
)
plt.title("Country-Specific vs Global (Random Forest)")
plt.ylabel("$R^2$")
plt.xlabel("")
plt.ylim(0, 1.05)
plt.legend(title="")
plt.tight_layout()
rf_plot_path = os.path.join(model_dir, "country_vs_global_RF.png")
plt.savefig(rf_plot_path, dpi=300)
plt.show()

# XGBoost comparison plot
plt.figure(figsize=(12, 4))
sns.barplot(
    data=xgb_comp_df,
    x="country",
    y="r2",
    hue="model",
)
plt.title("Country-Specific vs Global (XGBoost)")
plt.ylabel("$R^2$")
plt.xlabel("")
plt.ylim(0, 1.05)
plt.legend(title="")
plt.tight_layout()
xgb_plot_path = os.path.join(model_dir, "country_vs_global_XGB.png")
plt.savefig(xgb_plot_path, dpi=300)
plt.show()

print("Saved RF plot to:", rf_plot_path)
print("Saved XGB plot to:", xgb_plot_path)
```
