```python
# %%
# ==========================================================
# LSMS + EE_harvest: RF & XGB with Train/Test + Country Models
# ==========================================================

import os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from tqdm import tqdm

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GroupKFold, ParameterGrid
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

from xgboost import XGBRegressor

RANDOM_STATE = 42

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

print("Imports complete.")

# %%
# ----------------------------------------------------------
# 1. Paths & Loading
# ----------------------------------------------------------

LSMS_PATH = "/Users/jjburrell/Downloads/send/lsms_panel_allvars_2010_2024_MASTER.csv"
EE_PATH   = "/Users/jjburrell/Downloads/send/EE_harvest_ml.csv"

print("🔹 LSMS path:", LSMS_PATH)
print("🔹 EE path:", EE_PATH)

lsms_df = pd.read_csv(LSMS_PATH)
ee_df   = pd.read_csv(EE_PATH)

print("LSMS shape:", lsms_df.shape)
print("LSMS columns:", lsms_df.columns.tolist())
display(lsms_df.head())

print("\nEE shape:", ee_df.shape)
print("EE columns:", ee_df.columns.tolist())
display(ee_df.head())

# %%
# ----------------------------------------------------------
# 2. Build target from LSMS (yield if present, else NDVI)
# ----------------------------------------------------------

# LSMS columns given: ea_id_obs, ea_id_merge, lat_modified, lon_modified, year, month, ym,
# precip, tmax, tmin, soil_moist, NDVI, EVI, GCVI, GDD, KDD, source_file, date

id_keys = ["ea_id_obs", "ea_id_merge", "lat_modified", "lon_modified"]

# Detect target column
candidate_yields = ["yield_kg", "yield", "yield_ha", "yield_kg_ha"]
yield_col = next((c for c in candidate_yields if c in lsms_df.columns), None)

if yield_col is not None:
    target_raw = yield_col
    use_yield = True
    print(f"✅ Using {yield_col} as target.")
else:
    target_raw = "NDVI"
    use_yield = False
    if target_raw not in lsms_df.columns:
        raise ValueError("Neither yield_kg-like columns nor NDVI found in LSMS.")
    print("⚠️ No yield column found; using NDVI as target.")

# Basic cleaning
needed = id_keys + [target_raw]
lsms_df = lsms_df.dropna(subset=needed)

if use_yield:
    lsms_df = lsms_df[(lsms_df[target_raw] > 0) & (lsms_df[target_raw] < 15000)]

print("LSMS after cleaning:", lsms_df.shape)

# Aggregate per location (same spirit as original model)
agg_lsms = (
    lsms_df
    .groupby(id_keys, dropna=False)
    .agg(mean_target=(target_raw, "mean"))
    .reset_index()
)

if use_yield:
    agg_lsms["log_yield"] = np.log1p(agg_lsms["mean_target"])
    y_var = "log_yield"
else:
    y_var = "mean_target"

print("Aggregated LSMS shape:", agg_lsms.shape)
display(agg_lsms.head())

# %%
# ----------------------------------------------------------
# 3. Merge LSMS with EE (country/wave/season come from EE)
# ----------------------------------------------------------

# EE columns include: country, wave, season, ea_id_obs, ea_id_merge, lat_modified, lon_modified, ...

for k in id_keys:
    if k not in ee_df.columns:
        raise ValueError(f"EE is missing merge key: {k}")

df = pd.merge(
    agg_lsms,
    ee_df,
    on=id_keys,
    how="inner"
)

df = df.dropna()
print("✅ Merged LSMS + EE shape:", df.shape)
display(df.head())

# After merge, df now has 'country', 'wave', 'season' from EE
if not {"country", "wave", "season"}.issubset(df.columns):
    raise ValueError("Merged dataframe does not contain country/wave/season from EE.")

# %%
# ----------------------------------------------------------
# 4. Define X, y, groups, and Train/Test split by wave
# ----------------------------------------------------------

merge_vars = id_keys + ["country", "wave", "season"]

x_vars = [c for c in df.columns if c not in merge_vars + ["mean_target", "log_yield"]]
X_full = df[x_vars]
y_full = df[y_var]

print("Target variable:", y_var)
print("Number of features:", len(x_vars))
print("Example features:", x_vars[:15])

# Train/test: last wave = test
max_wave = int(df["wave"].max())
train_mask = df["wave"] < max_wave
test_mask  = df["wave"] == max_wave

X_train = X_full.loc[train_mask].copy()
y_train = y_full.loc[train_mask].copy()

X_test  = X_full.loc[test_mask].copy()
y_test  = y_full.loc[test_mask].copy()
meta_test = df.loc[test_mask].copy()

print(f"\nTrain waves: {sorted(df.loc[train_mask, 'wave'].unique())}")
print(f"Test wave: {max_wave}")
print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
print("Train countries:", df.loc[train_mask, "country"].unique())
print("Test countries:", df.loc[test_mask, "country"].unique())

# Spatial grouping for CV (per location)
groups_full = df["ea_id_merge"]
groups_train = groups_full.loc[train_mask].reset_index(drop=True)
print("GroupKFold groups prepared on ea_id_merge.")

# %%
# ----------------------------------------------------------
# 5. Hyperparameter grids and CV helper
# ----------------------------------------------------------

cv = GroupKFold(n_splits=5)

rf_param_grid = {
    "n_estimators": [300, 600],
    "max_depth": [12, 24, None],
    "max_features": [0.5, 0.8, 1.0],
    "min_samples_split": [2, 8],
    "min_samples_leaf": [1, 4],
    "bootstrap": [True],
}

xgb_param_grid = {
    "n_estimators": [400, 800],
    "max_depth": [4, 6, 8],
    "learning_rate": [0.03, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0],
    "reg_lambda": [1.0, 5.0],
}

rf_grid_list  = list(ParameterGrid(rf_param_grid))
xgb_grid_list = list(ParameterGrid(xgb_param_grid))

print(f"RF combos:  {len(rf_grid_list)}")
print(f"XGB combos: {len(xgb_grid_list)}")

def spatial_cv_tuning(
    model_name,
    model_class,
    base_kwargs,
    grid_list,
    X,
    y,
    groups,
    cv,
):
    best_r2 = -np.inf
    best_params = None

    for params in tqdm(grid_list, desc=f"Tuning {model_name}", ncols=100):
        r2_folds = []

        for tr_idx, va_idx in cv.split(X, y, groups):
            X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
            y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

            model = model_class(**base_kwargs, **params)
            model.fit(X_tr, y_tr)

            y_pred = model.predict(X_va)
            r2 = r2_score(y_va, y_pred)
            r2_folds.append(r2)

        r2_mean = float(np.mean(r2_folds))

        if r2_mean > best_r2:
            best_r2 = r2_mean
            best_params = params

    print(f"✅ {model_name} best CV R²={best_r2:.3f}")
    print("Best params:", best_params)
    return best_params, best_r2

def evaluate_model(model, X, y):
    y_pred = model.predict(X)
    r2 = r2_score(y, y_pred)
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    mae = mean_absolute_error(y, y_pred)
    return r2, rmse, mae, y_pred

# %%
# ----------------------------------------------------------
# 6. Tune + train GLOBAL RF & XGB (train set only)
# ----------------------------------------------------------

rf_base_kwargs = {
    "n_jobs": -1,
    "random_state": RANDOM_STATE,
}

xgb_base_kwargs = {
    "n_jobs": -1,
    "tree_method": "hist",
    "random_state": RANDOM_STATE,
}

print("\n🌍 Tuning GLOBAL Random Forest...")
best_rf_params, rf_cv_r2 = spatial_cv_tuning(
    model_name="GLOBAL_RF",
    model_class=RandomForestRegressor,
    base_kwargs=rf_base_kwargs,
    grid_list=rf_grid_list,
    X=X_train.reset_index(drop=True),
    y=y_train.reset_index(drop=True),
    groups=groups_train,
    cv=cv,
)

print("\n🌍 Tuning GLOBAL XGBoost...")
best_xgb_params, xgb_cv_r2 = spatial_cv_tuning(
    model_name="GLOBAL_XGB",
    model_class=XGBRegressor,
    base_kwargs=xgb_base_kwargs,
    grid_list=xgb_grid_list,
    X=X_train.reset_index(drop=True),
    y=y_train.reset_index(drop=True),
    groups=groups_train,
    cv=cv,
)

global_rf = RandomForestRegressor(**rf_base_kwargs, **best_rf_params)
global_rf.fit(X_train, y_train)

global_xgb = XGBRegressor(**xgb_base_kwargs, **best_xgb_params)
global_xgb.fit(X_train, y_train)

print("\n✅ Global RF and XGB trained.")

# %%
# ----------------------------------------------------------
# 7. Train country-specific RF & XGB (train set only)
# ----------------------------------------------------------

countries = sorted(df["country"].unique())
country_models_rf = {}
country_models_xgb = {}

for c in countries:
    train_mask_c = (df["country"] == c) & train_mask

    Xc_train = X_full.loc[train_mask_c]
    yc_train = y_full.loc[train_mask_c]

    print(f"\n=== Country {c} ===")
    print(f"  Train n = {len(Xc_train)}")

    if len(Xc_train) < 100:
        print("  ⚠️ Too few training samples, skipping country-specific models.")
        continue

    groups_c_train = df.loc[train_mask_c, "ea_id_merge"].reset_index(drop=True)

    print("  Tuning RF...")
    best_rf_c, _ = spatial_cv_tuning(
        model_name=f"RF_{c}",
        model_class=RandomForestRegressor,
        base_kwargs=rf_base_kwargs,
        grid_list=rf_grid_list,
        X=Xc_train.reset_index(drop=True),
        y=yc_train.reset_index(drop=True),
        groups=groups_c_train,
        cv=cv,
    )

    rf_c = RandomForestRegressor(**rf_base_kwargs, **best_rf_c)
    rf_c.fit(Xc_train, yc_train)
    country_models_rf[c] = rf_c

    print("  Tuning XGB...")
    best_xgb_c, _ = spatial_cv_tuning(
        model_name=f"XGB_{c}",
        model_class=XGBRegressor,
        base_kwargs=xgb_base_kwargs,
        grid_list=xgb_grid_list,
        X=Xc_train.reset_index(drop=True),
        y=yc_train.reset_index(drop=True),
        groups=groups_c_train,
        cv=cv,
    )

    xgb_c = XGBRegressor(**xgb_base_kwargs, **best_xgb_c)
    xgb_c.fit(Xc_train, yc_train)
    country_models_xgb[c] = xgb_c

print("\n✅ Finished training country-specific models.")

# %%
# ----------------------------------------------------------
# 8. Evaluate GLOBAL vs Country-Specific on TEST wave
# ----------------------------------------------------------

rows_rf = []
rows_xgb = []

for c in countries:
    test_mask_c = (df["country"] == c) & test_mask
    Xc_test = X_full.loc[test_mask_c]
    yc_test = y_full.loc[test_mask_c]

    print(f"\n>>> {c} test n = {len(Xc_test)}")
    if len(Xc_test) == 0:
        print("  No test data in held-out wave; skipping.")
        continue

    # Global
    r2_rf_glob, _, _, _ = evaluate_model(global_rf, Xc_test, yc_test)
    r2_xgb_glob, _, _, _ = evaluate_model(global_xgb, Xc_test, yc_test)

    rows_rf.append({"country": c, "model": "Global RF", "r2": r2_rf_glob})
    rows_xgb.append({"country": c, "model": "Global XGB", "r2": r2_xgb_glob})

    # Country-specific if available
    if c in country_models_rf:
        r2_rf_c, _, _, _ = evaluate_model(country_models_rf[c], Xc_test, yc_test)
        rows_rf.append({"country": c, "model": "Country-Specific RF", "r2": r2_rf_c})

    if c in country_models_xgb:
        r2_xgb_c, _, _, _ = evaluate_model(country_models_xgb[c], Xc_test, yc_test)
        rows_xgb.append({"country": c, "model": "Country-Specific XGB", "r2": r2_xgb_c})

rf_comp_df = pd.DataFrame(rows_rf)
xgb_comp_df = pd.DataFrame(rows_xgb)

print("\nRandom Forest comparison:")
display(rf_comp_df.sort_values(["country", "model"]))

print("\nXGBoost comparison:")
display(xgb_comp_df.sort_values(["country", "model"]))

# %%
# ----------------------------------------------------------
# 9. Plots: Country-Specific vs Global R² (like your figures)
# ----------------------------------------------------------

plt.figure(figsize=(12, 4))
sns.barplot(
    data=rf_comp_df,
    x="country",
    y="r2",
    hue="model",
)
plt.title("Country-Specific vs Global (Random Forest) — Test Wave")
plt.ylabel("$R^2$")
plt.xlabel("")
plt.ylim(0, 1.05)
plt.legend(title="")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 4))
sns.barplot(
    data=xgb_comp_df,
    x="country",
    y="r2",
    hue="model",
)
plt.title("Country-Specific vs Global (XGBoost) — Test Wave")
plt.ylabel("$R^2$")
plt.xlabel("")
plt.ylim(0, 1.05)
plt.legend(title="")
plt.tight_layout()
plt.show()
```
