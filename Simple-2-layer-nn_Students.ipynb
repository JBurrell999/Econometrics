## 1. Comparison of the different activation functions:

## (https://en.wikipedia.org/wiki/Activation_function, and there are more)

Logistic (a.k.a. Soft step, sigmoid) Activation: $f(x)=\frac{1}{1+e^{-x}}$, $\,f'(x)=f(x)(1-f(x))$,    ($0$, $1$)

TanH Activation: $f(x)=\tanh(x)=\frac{2}{1+e^{-2x}}-1$,	$\,f'(x)=1-f(x)^2$,   ($-1$, $1$)

```python
%matplotlib inline
# All imports
from random import choice
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(formatter={'float': '{:.4f}'.format})
```

```python
def sigmoid(x):
    '''
    Used in forward propogration
    The logistic function as the sigmoid
    '''

    return 1.0/(1.0 + np.exp(-x))

def sigmoid_pr(z):
    '''
    Used in backprop
    derivative of the logistic function
    '''
    return z * (1 - z)
```

## Breakout Excercise -- Write a function forwardprop(X, w0, w1) that performs the forward propagation

   - ## X is the input array
   - ## w0 are the weights in the first layer (it should be a 2x2 array...and think dot product)
   - ## w1 are the weights for the second layer (it has 2 weights)
   - ## It should return z, b, A, a
   - ## Test it with the input values and weights given in lecture.  And then print out the following, and make sure they agree to at least 7th decimal place:
   
            X = [0.35, 0.9]  
 
            w0:
             [[0.1000 0.4000]
             [0.8000 0.6000]]
            w1:
             [0.3000 0.9000]
            a: [0.7550 0.6800]
            A: [0.6803 0.6637]
            b: 0.801444986674
            output (z) of nn: 0.690283492908 

```python
def forwardprop(x, w0, w1):
    # X = np.append(X, 1) # add bias term
    # z1 = x @ w0
    z1 = np.dot(x, w0)
    a1 = sigmoid(z1)
    # z2 = a1 @ w1
    z2 = np.dot(a1, w1)
    output = sigmoid(z2)
    return z1, a1, z2, output
```

```python
X = np.array([0.35, 0.9])
print(X.shape)
w0 = np.array([[0.1000, 0.4000], [0.8000, 0.6000]])
print(w0.shape)
w1 = np.array([0.3000, 0.9000])
a, A, b, output = forwardprop(X, w0, w1)
print(f"a: {a}")
print(f"A: {A}")
print(f"b: {b}")
print(f"outputc(z) of nn: {output}")
```

## The inner (dot) product and the outer product of two vectors

```python
C = np.atleast_2d(np.array([1, 2]))
D = np.atleast_2d(np.array([3, 4]))

# print(A)
# print(B.T)

inner_prod = np.dot(C, D.T)
print("Inner product -- tensor rank goes down by 1") # 1D -> scalar
print("first matrix:", C.shape)
print("second matrix:", D.T.shape)

print('inner product:', inner_prod, inner_prod.shape)

print()

print("Outer Product -- tensor rank goes up by 1") # 1D -> 2D
print("first matrix:", D.T.shape)
print("second matrix:", C.shape)

outer_prod = np.dot(D.T, C)

print('C transposed:\n', C.T)
print('D:', D)
print('outer product:\n', outer_prod)

print()

print('Element by element multiplication -- preserves the the rank of the tensor')
print('element by element multiplication:', C * D)
```

## Partial Derivatives of $P$ w.r.t. $W^{(0)}_{ij}$, where $i, j = 0$ or $1$.

## Remember: The index of $x_i$ refers to the inputs and the index of $\mathrm{delta1}_j$ refers to the neurons in layer 1.

## $\frac {\partial P} {\partial W^{(0)}_{00}}$ = $x_0$ $\mathrm{delta1}_0$

## $\frac {\partial P} {\partial W^{(0)}_{01}}$ = $x_0$ $\mathrm{delta1}_1$

## $\frac {\partial P} {\partial W^{(0)}_{10}}$ = $x_1$ $\mathrm{delta1}_0$

## $\frac {\partial P} {\partial W^{(0)}_{11}}$ = $x_1$ $\mathrm{delta1}_1$

## This is the *outer* product of $\vec{x}$ and $\overrightarrow{\mathrm{delta1}}$.  This makes sense because there are four weights to be adjusted.

## Breakout Excercise -- Write a function 

`backprop(X, y, z, w0, w1)`

## that performs the backward propagation

   - ## The function should return the updated weights, w0 and w1
   - ## y is the target (= 0.5 in our case)
   - ## Define delta1 as we talked about in lecture
   - ## Also for convenience and clarity, define delta_w1 = delta1 * A.  This is essentially how w1 should be adjusted (with alpha = 1).
   - ## Then define delta1 as we talked about in class -- think about the best way to do it 
   - ## Define a delta_w0.   This is essentially how w0 should be adjusted (with alpha = 1). (*Hint*: Think matrix multiplication.)

```python
def backprop(X, y, z, w0, w1, verbose=False):
    g2_prime = z * (1 - z)
    delta1 = (y - z) * g2_prime
    delta_w1 = delta1 * A
    w1 = w1 + delta_w1
    delta0 = np.atleast_2d(sigmoid_pr(A) * delta1 * w1)
    # delta0 dimension: (hidden_dim, 1)
    # X dimension: (1, input_dim)
    # We want the delta weights to be in the same dimension of original weight, so we must do X.T @ delta0
    delta_w0 = np.atleast_2d(X).T @ delta0
    w0 += delta_w0

    if verbose:
        print(delta1)
        print(delta_w1)
        print(w1)
        print(delta0)
        print(delta_w0)
        print(w0)
```
