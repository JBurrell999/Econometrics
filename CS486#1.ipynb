```python
#N = Negative images
#P = positive images
#p = precision
#r = recall

N = 6300   
P = 190    
p = 0.83   
r = 0.87

#Formulas TP, FP, and FN as given 
TP = r * P
FN = P - TP
FP = TP * (1/p - 1)

#Formulas for TRP and FPR
TPR = TP / P
FPR = FP / N

print(TP, FN, FP, TPR, FPR)
```

Problem B:

```python
total = 10_000
trueRate = 1 / 10_000

expAcu = total * trueRate          
expAcuNeg = total - expAcu    

expTFlag = TPR * expAcu
expFFlag = FPR * expAcuNeg
expFTotal = expTFlag + expFFlag

inspectPerTrue = expFTotal / expTFlag
print(expTFlag, expFFlag, expFTotal, inspectPerTrue)
```
