# A note on whitespace

If were to write this loop in R it would like something like this:

```R
for(i in seq_along(x_values)){
    # do stuff
}
```

In Python we don't need `()` and `{}` in our loop because whitespace -- tabs, or four spaces -- essentially handle them for us. 

By removing clutter and **forcing** indentation, Python is often more readable than other languages. 

This is by design: one of the main insights from the developer of Python, Guido Van Rossum, is that code is more often **read** than written. 

So keep in mind that in any program that ends with `:`, like `for x in x_values:`, the following lines need to be indented, otherwise Python will throw an error.

## Commenting functions

We know we can comment individual lines of code with `#`. 

Functions have another commenting system called docstrings. 

You add them at the top of the function, like so:

````python
def square_list(X):
    """
    This function squares each element of a list.
    
    Input: X, a list
    
    Output: x_squared, a list of each element of X squared
    """
    x_squared = []
    for x in X:
        x_squared.append(x**2)
    return(x_squared)
````

# Important property of list object 

**The same is true about Numpy and Pandas, which we will discuss in the future**

Sometimes you want to copy a list. 

Mind that you copy it the right way. 

Say we define `some_list = [2,1]`:

We decide to make a copy to `new_list` using assignment `=`:

Then we decide to sort the values in `new_list`:

Our original list should be unaffected, right?

No. That is because copying lists with `=` does not actually create a new copy. 

Instead it creates a **pointer** or reference. 

`new_list` is just a reference to `some_list`. 

That means when we change one list we change the other. 

This happens because lists are **mutable** -- they can change. 

## Shallow copies

To create a true copy we need to use `.copy()`.

This creates a "shallow" copy of the original list:

Now when we change `another_new_list`:

our original list is unaffected:

## Deep copies

Shallow copies are not recursive. Basically that means they are not fully independent copies. 

They work fine if our list is a vector. 

What if we have a matrix -- a nested list?

````python
mat =[[1, 2], [3, 4]]
mat
````

````python
new_mat = mat.copy()
new_mat
````

````python
new_mat[0][0] = 10
new_mat
````

This doesn't work because `copy()` (which gives the same result as `list()`) only copies the outermost list (the "parent" list) and makes references to the inner lists (the "child" lists). 

To make a truly independent copy we need to use `deepcopy()` from the `copy` module:

````python
mat = [[1,2], [3,4]]
mat
````

````python
import copy
new_mat = copy.deepcopy(mat)
new_mat
````

````python
new_mat[0][0] = 10
new_mat
````
