```python
'''
The difference between the two sets of filters: 

original Sobel -- does trangular filtering perpendicular to the deriviative direction

alternative -- does top hat filter in the direction perpendicular to the derviative direction

Keep in mind that, e.g., in the horizontal edge filter, you have to make a choice for what to do for the first (and third) row

If you use [1, 1, 1] that amounts to a tophat filter in the horizontal direction
If you use [1, 2, 1] that amounts to a triangular filter in the horiztion direction

Whatever choice you make, it amounts to some kind of filter in the horizontal direction

Between the "standard" sobel filers and the alternative filters, which is better?


Hard to tell -- but this is the beauty of neural network: you don't have to try to be really clever
and "engineer" the best filters (which was the thinking priort to ~2010); rather a nn can "learn" 
and figure out the best filter for a particular purpose.


'''

# Apply the Sobel filter for horizontal and vertical edge detection
edges_horizontal = convolve(image, sobel_horizontal)
edges_vertical = convolve(image, sobel_vertical)


# alt_hor = np.array([[-1, -1, -1], 
#                     [ 0,  0,  0], 
#                     [ 1,  1,  1]])
# alt_ver = np.array([[-1, 0, 1], 
#                     [-1, 0, 1], 
#                     [-1, 0, 1]])

# edges_horizontal = convolve(image, alt_hor)
# edges_vertical = convolve(image, alt_ver)
 
    
fig, (ax_o, ax_h, ax_v) = plt.subplots(1, 3, figsize=(15, 5))
ax_o.imshow(image, cmap='gray')
ax_h.imshow(edges_horizontal, cmap='gray')
ax_v.imshow(edges_vertical, cmap='gray')
```
