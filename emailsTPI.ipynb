```python
# TPI Bounced Email Lookup Helper (Google + SerpAPI)

import pandas as pd
import requests

# === CONFIG ===
SERP_API_KEY = "YOUR_SERPAPI_KEY"  # Replace with your key
SEARCH_ENGINE = "google"

# Load Excel with name + company info
file_path = "/Users/jjburrell/Downloads/contactsTPI.xlsx"
df = pd.read_excel(file_path)

# Filter only entries with available names or orgs
df = df[df['First name'].notna() | df['Company'].notna()].copy()

# Create search query function
def make_query(row):
    name = f"{row['First name']} {row['Last name']}" if pd.notna(row['First name']) else ""
    company = row['Company'] if pd.notna(row['Company']) else ""
    domain = row['Domain'] if pd.notna(row['Domain']) else ""
    query = f'"{name}" "{company}" email contact'
    if domain:
        query += f" site:{domain}"
    return query

df['Search Query'] = df.apply(make_query, axis=1)

def google_search(query, api_key):
    url = "https://serpapi.com/search"
    params = {
        "engine": SEARCH_ENGINE,
        "q": query,
        "api_key": api_key,
        "num": 3  # top 3 resultsp
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        return {}

df['Top URLs'] = ""
df['Snippets'] = ""

for i, row in df.head(10).iterrows():  # limit for demo
    result = google_search(row['Search Query'], SERP_API_KEY)
    if 'organic_results' in result:
        urls = []
        snippets = []
        for r in result['organic_results']:
            urls.append(r.get('link', ''))
            snippets.append(r.get('snippet', ''))
        df.at[i, 'Top URLs'] = " | ".join(urls)
        df.at[i, 'Snippets'] = " | ".join(snippets)

df.to_excel("updated_bounces_with_search.xlsx", index=False)
print("✅ Search results saved to updated_bounces_with_search.xlsx")

import pandas as pd
import requests

# === CONFIG ===
SERP_API_KEY = "YOUR_SERPAPI_KEY"  
SEARCH_ENGINE = "google"

file_path = "/Users/jjburrell/Downloads/contactsTPI.xlsx"
df = pd.read_excel(file_path)

df = df[df['First name'].notna() | df['Company'].notna()].copy()

def make_query(row):
    name = f"{row['First name']} {row['Last name']}" if pd.notna(row['First name']) else ""
    company = row['Company'] if pd.notna(row['Company']) else ""
    domain = row['Domain'] if pd.notna(row['Domain']) else ""
    query = f'"{name}" "{company}" email contact'
    if domain:
        query += f" site:{domain}"
    return query

df['Search Query'] = df.apply(make_query, axis=1)

def google_search(query, api_key):
    url = "https://serpapi.com/search"
    params = {
        "engine": SEARCH_ENGINE,
        "q": query,
        "api_key": api_key,
        "num": 3  
    }
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()
    else:
        return {}

# Add empty cols for manual review
df['Top URLs'] = ""
df['Snippets'] = ""

# Run lookup for first N rows for demo
for i, row in df.head(10).iterrows():  # limit for demo
    result = google_search(row['Search Query'], SERP_API_KEY)
    if 'organic_results' in result:
        urls = []
        snippets = []
        for r in result['organic_results']:
            urls.append(r.get('link', ''))
            snippets.append(r.get('snippet', ''))
        df.at[i, 'Top URLs'] = " | ".join(urls)
        df.at[i, 'Snippets'] = " | ".join(snippets)

# Save updated sheet for manual review
df.to_excel("/Users/jjburrell/Downloads/updated_bounces_with_search.xlsx", index=False)
print("✅ Search results saved to updated_bounces_with_search.xlsx")
```
